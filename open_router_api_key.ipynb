````python
# ---------- Node 3: Execute SQL ----------
def execute_query(state: State):
    print("Executing node 3:\n")
    query = state.get("query", "")
    check = state.get("check", 0)
    retries = state.get("retries", 0)
    max_retries = state.get("max_retries", 3)
    #check_done=state.get("check_done",0)

    print("[execute_query] Executing:", query)

    try:
        with engine.connect() as conn:
            result = conn.execute(text(query))
            # Use mappings() to get dict-like rows in SQLAlchemy 1.4+
            rows = result.mappings().all()
    except Exception as e:
        print("[execute_query] SQL execution error:", e)
        # increment counters and signal failure by setting check>0
        return {"answer": [], "check": check + 1, "retries": retries + 1}

    # If query ran but returned empty rows → treat as failure (per your spec)
    if  not rows or len(rows) == 0:
        print("[execute_query] Query executed but returned 0 rows.")
        return {"answer": [], "check": check + 1, "retries": retries + 1}
        #return {"answer": [], "check_done": check_done + 1,"retries":retries}

    # Success: return rows and reset check
    print(f"[execute_query] Success — {len(rows)} rows.")
    return {"answer": rows, "check": 0, "retries": retries}
    
# ---------- Node 4: check_result (regenerate query) ----------
````

````python
def route_execute(state: State) -> str:
    print("Executing Routing Node 3:\n")
    """
    Should return one of the keys in mapping passed to add_conditional_edges:
    - 'success' when we have a valid answer
    - 'retry' when we need to re-generate query
    - 'max_retries' when retry limit reached
    """
    # check for retries / success
    retries = state.get("retries", 0)
    max_retries = state.get("max_retries", 3)
    answer = state.get("answer", [])
    check = state.get("check", 0)
    #check_done = state.get("check_done")

    if (answer and len(answer) > 0 and check == 0): #or check_done > 0:
        return "success"

    if retries >= max_retries:
        return "max_retries"

    # otherwise we need to retry / correct the query
    return "retry"
````

````python
def check_result(state: State):
    print("Executing node 4:\n")
    """
    Called when route_execute returned 'retry'. The node asks llm to fix the query
    based on the schema and user question, returns a new 'query' to be executed.
    """
    query = state.get("query", "")
    question = state.get("question", "")

    prompt=f"""The previous SQL query was incorrect or returned no rows:\n{query}\n
              User question: {question}\n
              Database schema: {db_schema}\n
              Please output ONLY a corrected SELECT SQL query (no explanation)."""
    
    response = requests.post(
    url="https://openrouter.ai/api/v1/chat/completions",
    headers={
    "Authorization": "Bearer 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",
    "Content-Type": "application/json",},
    data=json.dumps({
    #"model": "mistralai/devstral-2512:free",
        "model": "nex-agi/deepseek-v3.1-nex-n1:free",
    "messages": [  { "role": "system", "content": system_prompt},{ "role": "user", "content": prompt}   ]   })  )
    new_q = response.json()["choices"][0]["message"]["content"]
    #new_q = llm.invoke(prompt).content.strip()
    new_q = new_q.replace("```sql", "").replace("```", "").strip()

    print("[check_result] New query generated:\n", new_q)
    # set check back to 0 because we are trying again; do not reset retries
    return {"query": new_q, "check": 0}
````

````python
# ---------- Node 5: handle_max_retries (optional) ----------
def handle_max_retries(state: State):
    print("Executing 5th node\n")
    # This node will be reached when retry limit is exceeded.
    # You can either end with an error, return None/empty answer or set a flag.
    print("[handle_max_retries] Reached maximum retries. Stopping.")
    return {"answer": [], "check": 1}
````

````python
# ---------- Build the graph with conditional routing ----------
graph = StateGraph(State)
# graph.add_node("check_input", check_input)
graph.add_node("generate_query", generate_query)
#graph.add_node("extract_query", extract_query)
graph.add_node("execute_query", execute_query)
graph.add_node("check_result", check_result)
graph.add_node("handle_max_retries", handle_max_retries)

graph.set_entry_point("generate_query")

# linear initial flow
#graph.add_edge("generate_query", "extract_query")
graph.add_edge("generate_query", "execute_query")
# extract → execute
#graph.add_edge("extract_query", "execute_query")

# conditional routing from execute_query:
# - on success -> END
# - on retry -> go to check_result (which will generate new query)
# - on max_retries -> go to handle_max_retries (or END)
graph.add_conditional_edges(
    "execute_query",
    route_execute,
    {
        "success": END,
        "retry": "check_result",
        "max_retries": "handle_max_retries"
    }
)

# After check_result produces a new query, go directly to execute_query to try it.
graph.add_edge("check_result", "execute_query")

# handle_max_retries -> END (we could also route elsewhere)
graph.add_edge("handle_max_retries", END)
app = graph.compile()
````
