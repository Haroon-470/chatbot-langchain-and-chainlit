```python
from sqlalchemy import create_engine

DB_USER = "muhammad-haroon"
DB_PASS = "haroon123"
DB_HOST = "localhost"      # or server IP
DB_PORT = "5432"
DB_NAME = "euro_parts_data"

engine = create_engine(
    f"postgresql+psycopg2://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
```

```python
engine
```

```python
import pandas as pd
from tqdm import tqdm

TABLE_NAME = "product_product"
#CHUNK_SIZE = 100_000   # adjust if PC is very slow

query = f"SELECT * FROM {TABLE_NAME}"

chunk_iter = pd.read_sql_query(query, engine)

for i, chunk in enumerate(tqdm(chunk_iter)):
    file_name = f"{TABLE_NAME}_part_{i}.csv"
    chunk.to_csv(file_name, index=False)
```

```python
TABLE_NAME = "september_month"

query = f"SELECT * FROM {TABLE_NAME}"

# ---- Read full table into DataFrame ----
df = pd.read_sql_query(query, engine)

# ---- Save to CSV ----
df.to_csv(f"{TABLE_NAME}.csv", index=False)

print("✅ Export completed successfully")
```
